### Find all the confirmed eruptions of a given volcano (from the start year)

```python
volcano_name = 'Fournaise'
df_volcano = data_recent[(data_recent['Volcano Name'] == 'Fournaise, Piton de la')]
df_volcano = df_volcano.sort_values(by='Start Year')
df_volcano
```

```python
list_dates= []

for index, eruption in df_volcano.iterrows() :

    year = eruption['Start Year']
    month = int(eruption['Start Month'])
    day = int(eruption['Start Month'])

    if len(str(month)) == 1 and len(str(month)) ==1 :
        list_dates.append(f'{year}-0{month}-0{day}T00:00:00')
    elif len(str(month)) == 1 :
        list_dates.append(f'{year}-0{month}-{day}T00:00:00')
    elif len(str(day)) == 1 :
        list_dates.append(f'{year}-{month}-0{day}T00:00:00')
    else :
        list_dates.append(f'{year}-{month}-{day}T00:00:00')
```

```python
with open(f'./{volcano_name}.txt', 'w+') as f:

    # write elements of list
    for items in list_dates:
        f.write('%s\n' %items)
    print("File written successfully")
# close the file
f.close()
```
