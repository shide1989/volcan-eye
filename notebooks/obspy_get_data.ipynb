```python
# by 4-month chunks

def compute_date(year, part) :

    var_year = year

    if part == 1 :
    ####### First part of the year #########

        var_start_month = 1 # January
        var_end_month = 4 # April

        var_start_day = '01'
        var_end_day = 30


        if len(str(var_start_month)) == 2 :
            var_start_month_str = str(var_start_month)
        else :
            var_start_month_str = '0' + str(var_start_month)


        if len(str(var_end_month)) == 2 :
            var_end_month_str = str(var_end_month)
        else :
            var_end_month_str = '0' + str(var_end_month)


        start_date = f'{var_year}-{var_start_month_str}-{var_start_day}T00:00:00'

        end_date = f'{var_year}-{var_end_month_str}-{var_end_day}T00:00:00'

        print(start_date)
        print(end_date)

        return start_date, end_date


    elif part == 2 :
    ####### Second part of the year #########

        var_start_month = 5 # May
        var_end_month = 8 # August

        var_start_day = '01'
        var_end_day = 31


        if len(str(var_start_month)) == 2 :
            var_start_month_str = str(var_start_month)
        else :
            var_start_month_str = '0' + str(var_start_month)


        if len(str(var_end_month)) == 2 :
            var_end_month_str = str(var_end_month)
        else :
            var_end_month_str = '0' + str(var_end_month)


        start_date = f'{var_year}-{var_start_month_str}-{var_start_day}T00:00:00'

        end_date = f'{var_year}-{var_end_month_str}-{var_end_day}T00:00:00'

        print(start_date)
        print(end_date)

        return start_date, end_date


    else :
    ####### Third part of the year #########

        var_start_month = 9 # September
        var_end_month = 12 # December

        var_start_day = '01'
        var_end_day = 31


        if len(str(var_start_month)) == 2 :
            var_start_month_str = str(var_start_month)
        else :
            var_start_month_str = '0' + str(var_start_month)


        if len(str(var_end_month)) == 2 :
            var_end_month_str = str(var_end_month)
        else :
            var_end_month_str = '0' + str(var_end_month)


        start_date = f'{var_year}-{var_start_month_str}-{var_start_day}T00:00:00'

        end_date = f'{var_year}-{var_end_month_str}-{var_end_day}T00:00:00'

        print(start_date)
        print(end_date)

        return start_date, end_date
```

```python
def get_data(start_date, end_date, path):

    # get the stream
    st = client.get_waveforms(
        network="G",           # code réseau
        station="RER",         # code station
        location="*",          # "*" = tous les codes de localisation, ou "00", "" etc.
        channel="BHZ",         # canal : BHZ = large bande, composante verticale
        starttime=UTCDateTime(start_date),
        endtime=UTCDateTime(end_date)
    )

    # get the instrument info
    inv = client.get_stations(network="G", station="RER", level="response")


    # local save
    st.write(f"{path}{start_date}_{end_date}.mseed", format="MSEED")
    inv.write(f"{path}{start_date}_{end_date}.xml", format="STATIONXML")

    return(st, inv)
```

```python
# Initialization
from obspy import Stream

path_data = './data_Fournaise_G_RER_BHZ/'

stream_concat = Stream() # empty stream

var_year = 2001 # start year for the loop
```

```python
while var_year <= 2002 :

    # First part of the year
    start, end = compute_date(var_year, part = 1)
    st, inv = get_data(start, end, path_data)
    stream_concat = stream_concat + st

    # Second part of the year
    start, end = compute_date(var_year, part = 2)
    st, inv = get_data(start, end, path_data)
    stream_concat = stream_concat + st

    # Third part of the year
    start, end = compute_date(var_year, part = 3)
    st, inv = get_data(start, end, path_data)
    stream_concat = stream_concat + st


    var_year +=1
```
